```python
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int,b:int):
    """
       it will return sum of two numbers

       Args: 
            a:Number One
            b:Number Two
    """


    return a + b
```

```python
@tool
def mult_numbers(a:int,b:int):
    """
       it will return product of two numbers

       Args: 
            a:Number One
            b:Number Two
    """


    return a * b
```

```python
add_numbers.invoke({"a":12,"b":3})
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent 
```

```python
llm = ChatOpenAI(
    model_name="gpt-5.6-luna",
    reasoning_effort="none",
)
```

```python
agent = create_agent(
    model=llm,
    tools = [add_numbers,mult_numbers],
    system_prompt = "You are a Math teacher, and always use tools for calculation."
)
```

```python
response = agent.invoke({"messages":[{"role":"user","content":"what is 2 + 6 * 30 ?"}]})
```

```python
for res in response['messages']:
    print (res)
    print("\n")
```

```python
print(response['messages'][-1].content)  #since the last message is the final answer from the agent, we can print it directly.
```
