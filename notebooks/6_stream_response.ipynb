```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm=ChatOpenAI(model="gpt-5.6-luna",streaming = True)
```

```python
question = "What is GPT meaning?"
```

```python
res=llm.stream(question)
for chunk in res:
    print(chunk.content,end= "")
```
