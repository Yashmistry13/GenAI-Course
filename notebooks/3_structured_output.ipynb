```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model ="gpt-5.6-luna")
```

## Problem

```python
text = "Hello, My name is Yash Mistry,"\
"Email is yash.mistry13@gmail.com and my age is 21"
```

```python
res= llm.invoke(f"please give me only name, email and age from this text :{text}")
```

```python
print(res.content)
```

## Solution - Structured Output

```python
from pydantic import BaseModel,Field ## imp library for defining data structures 
from typing import List

class ResponseStructure(BaseModel):
    name: str = Field(description ="Complete Name")
    email: str = Field(description="Email Address")
    age: int = Field(description="Age")

structured_llm=llm.with_structured_output(ResponseStructure)
```

```python
res = structured_llm.invoke(f"please give me only name, email and age from this text :{text}")
print(res)
```

```python
class Movies(BaseModel):
    title: str = Field(description ="Movie Name")
    year: int = Field(description="Release Year")
    rating: float = Field(description="IMDB Rating")
    genre: str = Field(description ="Movie Genre")

class AllMovies(BaseModel):
    movies:List[Movies] 

movie_llm = llm.with_structured_output(AllMovies)
```

```python
print(movie_llm.invoke("Give me top 2 Movie Recommendation from suspense genre that has won Oscar award"))
```
