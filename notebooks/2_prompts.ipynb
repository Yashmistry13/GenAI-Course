### Prompts Template - Dynamic Prompts 

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
llm = ChatOpenAI(model="gpt-5.6-luna")
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a Translator and translate text in {language}"},
    {"role":"user","content":"{query}"},])
```

```python
final_prompt = prompts.format(language="German", query="Lovely Nice to meet you!")
res=llm.invoke(final_prompt)
```

```python
print(res.content)
```

### Chains - Runnables

```python
chains= prompts | llm  ### directly chaining the prompt and llm together
# chains now becomes an invokable and runnable object 
```

```python
chains 
```

```python
res=chains.invoke({"language": "Portuguese", "query": "I Love you!"})
print(res.content)
```

```python
from langchain_core.output_parsers import StrOutputParser 
### Directly converts the output to string and returns it. 
# no need to use res.content to get the string output. 
```

```python
out=StrOutputParser()
```

```python
chains = prompts | llm | out
```

```python
res=chains.invoke({"language": "Bhojpuri", "query": "I Love langchain!"})
print(res)
```
